```python
import ollama
import json

def get_logits(prompt, model="llama3.2"):
    """
    Generates text using the specified model and returns logits.

    Args:
        prompt (str): The input prompt for text generation.
        model (str, optional): The name of the Ollama model to use. Defaults to "llama2".

    Returns:
        dict: A dictionary containing the generated text and logits.
              Returns None if an error occurs.
    """
    try:
        response = ollama.chat(model=model, messages=[{'role': 'user', 'content': prompt}], options={"raw": True}, stream=False)
        
        # Parse the raw JSON response
        print(response)
        raw_content = response['message']['content']
        json_content = json.loads(raw_content)

        # Extract the text and logits
        text = json_content["content"]
        logits = json_content["logits"]
        
        return {"text": text, "logits": logits}

    except Exception as e:
        print(f"An error occurred: {e}")
        return None


if __name__ == "__main__":
    prompt_text = "Tell me a short story about a cat."
    result = get_logits(prompt_text)

    if result:
        print("Generated Text:")
        print(result["text"])
        print("\nLogits:")
        print(result["logits"])
    else:
        print("Failed to get logits.")
```
